```python
my_figure=plt.figure()
my_axes=my_figure.add_axes([0.1,0.1,0.6,0.5])
my_axes.plot(numpy_array1,numpy_array2,"g")
my_axes.set_xlabel("X Axis")
my_axes.set_ylabel("Y Axis")
my_axes.set_title("Graph Title")
plt.show()
```

```python
new_figure=plt.figure(dpi=100)
new_axes=new_figure.add_axes([0.1,0.1,0.9,0.9])
new_axes.plot(numpy_array1,numpy_array1 **2 , label="Numpy Array *2")
new_axes.plot(numpy_array1,numpy_array1 **4 , label="Numpy Array *4")
new_axes.legend(loc=6)
plt.show()
```
