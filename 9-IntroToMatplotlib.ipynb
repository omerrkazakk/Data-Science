```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
%matplotlib inline
```

```python
age_list=[10,20,30,40,40,40,50,50,55,60,62,75]
```

```python
len(age_list)
```

```python
weight_list=[30,40,50,55,60,61,67,72,78,82,87,95]
```

```python
len(weight_list)
```

```python
plt.plot(age_list , weight_list , "r") #r --> red y -->yellow g-->green
plt.xlabel("Age")
plt.ylabel("Weight")
plt.title("Age vs Weight")
plt.show()
```
