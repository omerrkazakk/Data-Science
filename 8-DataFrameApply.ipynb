```python
def salary_category(salary):
    if salary <50000:
        return "Low"
    elif 50000<= salary <=80000:
        return "Medium"
    else:
        return "High"
    
```

```python
df_apply["Salary_Category"]=df_apply["Salary"].apply(salary_category)
```

```python
df_apply
```

```python

def experience_score(experience):
    if experience >10:
        return 1
    else:
        return 0

```

```python
df_apply["Adjusted"]=df_apply["Experience"].apply(experience_score)
```

```python
df_apply
```

```python
df_apply["New Score"]= df_apply["Performance_Score"] + df_apply["Adjusted"]
```

```python
df_apply
```

```python
def adjusted_score(row):
    if row["Experience"]>10:
        return row ["Performance_Score"]+1
    else:
        return row ["Performance_Score"]
        
```

```python
df_apply["Adjusted_Score"]=df_apply.apply(adjusted_score , axis= 1)
```

```python
def department_info(row):
    if row["Department"]=="Marketing":
        return row["Salary"]+10000
    elif row["Department"]=="Sales":
        return row["Salary"] +5000
    elif row["Department"]=="IT":
        return row["Salary"] +15000
    else:
        return row["Salary"]
    
```

```python
df_apply["New Salary"]=df_apply.apply(department_info , axis=1)
```

```python
df_apply
```

```python
def bonus (row):
    if row["Performance_Score"]==5:
        return row["Salary"]+8000
    elif row["Performance_Score"]==4:
        return row["Salary"]+5000
    elif row["Performance_Score"]==3:
        return row["Salary"]+3000
    elif row["Performance_Score"]==2:
        return row["Salary"]+1000
    else:
        return row["Salary"]
```

```python
df_apply["Bonus"]=df_apply.apply(bonus , axis=1)
```

```python
def employee_level(row):
    if row["Experience"]<=5:
        return "Junior"
    elif row["Experience"]<=15:
        return "Mid"
    elif row["Experience"]>15:
        return "Senior"
    else:
        return "Not Found Experience"
```

```python
df_apply["Employee_Level"]=df_apply.apply(employee_level , axis=1)
```

```python
df_apply
```

```python


def update_salary (row):
    if row["Performance_Score"]==5:
        return row["Salary"] *1.5
    elif row["Performance_Score"]==4:
        return row["Salary"] * 1.4
    elif row["Performance_Score"]==3:
        return row["Salary"] * 1.3
    elif row["Performance_Score"]==2:
        return row["Salary"] * 1.2
    else:
        return row["Salary"]*1.1
```

```python
df_apply["Updated_Salary"]=df_apply.apply(update_salary, axis=1)
```
