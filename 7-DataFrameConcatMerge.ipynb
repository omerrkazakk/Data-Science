```python
# merge 
```

```python
merge_df1= pd.read_csv("7-merge_data1.csv")
```

```python
merge_df2=pd.read_csv("7-merge_data2.csv")
```

```python
merge_df1
```

```python
merge_df2
```

```python
# merge outer join 
```

```python
df_merged_outer= pd.merge(merge_df1,merge_df2 , on="Employee_ID", how="outer")
```

```python
df_merged_outer
```

```python
# merge left join
```

```python
df_left=pd.merge(merge_df1,merge_df2 , on="Employee_ID" ,how="left")
```

```python
df_left
```

```python
# merge right join 
```

```python
df_right= pd.merge(merge_df1,merge_df2 , on="Employee_ID" , how="right")
```

```python
df_right
```

```python
#merge inner join 
```

```python
df_inner= pd.merge(merge_df1,merge_df2 , on="Employee_ID" , how="inner")
```
