```python
import numpy as np
```

```python
import matplotlib.pyplot as plt 
```

```python
%matplotlib inline
```

```python
data1=np.linspace(0,15,20)
```

```python
data1
```

```python
data2=data1 ** 5
```

```python
data2
```

```python
plt.subplots()
```

```python
my_fig ,my_axes=plt.subplots()
my_axes.plot(data1,data2)
my_axes.plot(data2,data1)
plt.show()
```

```python
my_fig ,my_axes=plt.subplots()
my_axes.plot(data1,data2 , color="#0ffcf5",alpha=0.3)
my_axes.plot(data2,data1 ,color="#d80ffc")
plt.show()
```

```python
(new_figs , new_axes)=plt.subplots()
new_axes.plot(data1,data1+10, color="blue",linewidth=1.0)
new_axes.plot(data1,data1+20, color="yellow",linewidth=2.0)
new_axes.plot(data1,data1+30, color="#58559e",linestyle="-.")
new_axes.plot(data1,data1+40, color="#559e5d",linestyle=":")
new_axes.plot(data1,data1+50, color="#919e55",linestyle="--")
new_axes.plot(data1,data1+60, color="#000000",linestyle="--",marker="o",markersize=8,markerfacecolor="red")
new_axes.plot(data1,data1+70, color="#000000",linestyle="--",marker="+",markersize=10,)
plt.show()
```
