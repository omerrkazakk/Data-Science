```python
import pandas as np
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
#box plot
```

```python
data = np.array([5,7,9,10,13,20,24,28,33,41,57,59,60,76,82,88,93,158])
plt.figure(figsize=(6,5))
sns.boxplot(y=data)
plt.title("Box Plot")
plt.ylabel("Data Value")
#plt.grid(True)
plt.show()
```

```python
data = np.array([5,7,9,10,13,20,24,28,33,35,49,56,56,56,56,56,56,56,56,56,56,57,59,60,76,])
plt.figure(figsize=(6,5))
sns.boxplot(y=data)
plt.title("Box Plot")
plt.ylabel("Data Value")
plt.grid(True)
plt.show()
```

```python
df = sns.load_dataset("titanic")
```

```python
df.head()
```

```python
plt.subplot(1,2,1)
sns.boxplot(x="class",y="age",data=df)
plt.title("Class by Age")
plt.xlabel("Class")
plt.ylabel("Age")

plt.subplot(1,2,2)
sns.boxplot(x="class",y="fare",data=df)
plt.title("Class by Fare")
plt.xlabel("Class")
plt.ylabel("Fare")

plt.show()
```
